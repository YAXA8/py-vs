```python
import numpy as np

# تنظیم Seed برای بازتولیدپذیری نتایج تصادفی
np.random.seed(42)

# 1. np.array: تعریف ماتریس امتیاز ۴ کاربر به ۳ پلی‌لیست موسیقی
user_ratings = np.array([
    [4.5, 3.0, 5.0],
    [2.0, 1.5, 4.0],
    [5.0, 4.5, 4.5],
    [3.5, 4.0, 2.0]
])

# 2. np.zeros: ساخت بافر خالی برای ثبت لاگ خطاهای سیستم (ابعاد: 4 سرور، 3 وضعیت)
error_buffer = np.zeros((4, 3), dtype=int)

# 3. np.ones: بردار ضرایب اولیه بایاس (Bias weights)
bias_weights = np.ones((3,))

# 4. np.arange: تایم‌لاین شبیه‌سازی تماشای ویدیو (از ثانیه 0 تا 60 با گام 10 ثانیه)
timeline_sec = np.arange(0, 61, 10)

# 5. np.linspace: تولید 5 نقطه زمانی دقیق برای آنالیز فریم‌های کلیدی یک صحنه 2 ثانیه‌ای
frame_timestamps = np.linspace(0.0, 2.0, num=5)

# 6. np.eye: ماتریس همانی برای نگاشت هویت ژانرها به یکدیگر
genre_identity = np.eye(3)

print("Ratings Matrix:\n", user_ratings)
print("Timeline:", timeline_sec)
print("Frames:", frame_timestamps)
```

```python
# 7. np.reshape: تبدیل بردار خطی لاگ استریم به ماتریس (۲ سشن، ۳ کاربر)
raw_stream_logs = np.arange(1, 7)
session_matrix = np.reshape(raw_stream_logs, (2, 3))

# 8. np.transpose (یا .T): ترانهاد ماتریس برای تغییر دیدگاه از (کاربر × پلی‌لیست) به (پلی‌لیست × کاربر)
transposed_ratings = np.transpose(user_ratings)

# 9. np.concatenate: الحاق داده‌های ۲ کاربر جدید به ماتریس اصلی
new_users = np.array([[4.0, 4.0, 4.0], [1.0, 2.0, 3.0]])
all_users_matrix = np.concatenate((user_ratings, new_users), axis=0)

# 10. np.expand_dims: افزودن بُعد Batch Size برای آماده‌سازی ورودی مدل (تبدیل به 3D)
batched_features = np.expand_dims(all_users_matrix, axis=0) # Shape: (1, 6, 3)

# 11. np.squeeze: حذف ابعاد اضافی با طول 1
restored_shape = np.squeeze(batched_features)

print("Concatenated Shape:", all_users_matrix.shape)
print("Batched Shape:", batched_features.shape)
```

```python
# 12. np.mean: میانگین امتیاز داده شده به هر پلی‌لیست (axis=0 ستونی)
avg_playlist_scores = np.mean(user_ratings, axis=0)

# 13. np.std: محاسبه انحراف معیار برای سنجش میزان توافق یا اختلاف نظر کاربران
score_variance = np.std(user_ratings, axis=0)

# 14. np.sum: مجموع زمان صرف‌شده توسط هر کاربر (axis=1 سطری)
total_user_engagement = np.sum(user_ratings, axis=1)

# 15. np.min / np.max: حداقل و حداکثر امتیاز ثبت شده در کل پلتفرم
min_score = np.min(user_ratings)
max_score = np.max(user_ratings)

# 16. np.argmax: شناسایی ایندکس پرطرفدارترین پلی‌لیست بر اساس میانگین
top_playlist_index = np.argmax(avg_playlist_scores)

print(f"Average Scores: {avg_playlist_scores}")
print(f"Top Playlist Index: {top_playlist_index} (Score: {avg_playlist_scores[top_playlist_index]:.2f})")
```

```python
# پروفایل علاقه‌مندی سیستم بر اساس ژانر (وزن‌های تخصیص‌یافته به هر پلی‌لیست)
weights = np.array([0.4, 0.35, 0.25])

# 17. np.dot: ضرب ماتریسی برای محاسبه نمره نهایی شاخص رضایت (User Satisfaction Score)
satisfaction_scores = np.dot(user_ratings, weights)

# 18. np.linalg.norm: محاسبه نرم بردار (طول بردار) برای نرمال‌سازی پروفایل کاربر
user_1_norm = np.linalg.norm(user_ratings[0])

# 19. np.exp: تابع Softmax ساده‌سازی‌شده برای تبدیل امتیازات به توزیع احتمالاتی ترجیحات
exp_scores = np.exp(user_ratings[0] - np.max(user_ratings[0])) # بهینه از نظر عددی
genre_probabilities = exp_scores / np.sum(exp_scores)

# 20. np.clip: محدود کردن مقادیر خارج از محدوده (مثلاً فیلتر کردن نویزهای فراتر از بازه [1.0, 5.0])
normalized_clamped_ratings = np.clip(user_ratings, a_min=2.0, a_max=4.5)

print("Satisfaction Scores:", satisfaction_scores)
print("Genre Probabilities (User 1):", np.round(genre_probabilities, 3))
```

```python
# 21. np.where: برچسب‌گذاری کاربران به "علاقه‌مند به ژانر اکشن" در صورت امتیاز >= 4.0
action_genre_scores = user_ratings[:, 0]
user_tags = np.where(action_genre_scores >= 4.0, "Action Lover", "Casual")

# 22. np.unique: شناسایی مقادیر متمایز در دسته‌بندی‌ها
unique_tags = np.unique(user_tags)

# 23. np.isin: بررسی اینکه آیا دسته‌های خاصی در لیست خروجی وجود دارند یا خیر
is_vip_present = np.isin(user_tags, ["Action Lover"])

# 24. np.sort / np.argsort: رتبه‌بندی کاربران بر اساس شاخص رضایت (صعودی/نزولی)
ranking_order = np.argsort(satisfaction_scores)[::-1] # مرتب‌سازی ایندکس‌ها به صورت نزولی

# 25. np.all / np.any: بررسی منطقی کلی
has_any_flawless = np.any(user_ratings == 5.0)  # آیا حداقل یک امتیاز 5 داریم؟
are_all_passed = np.all(user_ratings >= 1.0)     # آیا همه امتیازها معتبرند؟

print("User Classifications:", user_tags)
print("Ranked User Indices (Best to Worst):", ranking_order)
print(f"Contains 5.0 score? {has_any_flawless} | All >= 1.0? {are_all_passed}")
```

```python
import numpy as np

# تنظیم مجدد seed
np.random.seed(101)

# 26. np.random.normal: شبیه‌سازی مدت زمان تماشای ویدیو بر حسب دقیقه (میانگین 45، انحراف معیار 10)
watch_durations = np.random.normal(loc=45.0, scale=10.0, size=(100,))

# 27. np.random.randint: شبیه‌سازی تعداد کلیک‌های روزانه کاربران (بین 1 تا 50 کلیک)
user_clicks = np.random.randint(low=1, high=51, size=(5, 4))

# 28. np.random.choice: نمونه‌گیری تصادفی و وزن‌دار از دسته‌های تخفیف اشتراک
subscription_tiers = np.array(["Free", "Standard", "Premium"])
weights = [0.6, 0.3, 0.1]
sampled_users = np.random.choice(subscription_tiers, size=8, p=weights)

# 29. np.random.shuffle: بُر زدن درون‌جایی داده‌ها برای تقسیم‌بندی به Train/Test
dataset_indices = np.arange(10)
np.random.shuffle(dataset_indices)

print("Sampled Subscriptions:", sampled_users)
print("Shuffled Indices:", dataset_indices)
```

```python
# 30. np.median: محاسبه میانه زمان استریم (مقاوم در برابر مقادیر پرت)
median_watch_time = np.median(watch_durations)

# 31. np.percentile: یافتن آستانه 90 درصد کاربران برتر (P90 Latency/Engagement)
p90_threshold = np.percentile(watch_durations, 90)

# 32. np.quantile: آستانه کوانتایل 0.25 و 0.75 برای تحلیل دامنه بین چارکی (IQR)
q25, q75 = np.quantile(watch_durations, [0.25, 0.75])

# 33. np.cumsum: محاسبه زمان انباشته (Cumulative Sum) استریم کاربر در طول هفته
daily_hours = np.array([1.5, 2.0, 0.5, 3.0, 4.2, 1.0, 2.8])
cumulative_watch = np.cumsum(daily_hours)

# 34. np.diff: محاسبه آهنگ تغییرات (شیب مصرف) روزانه
day_to_day_change = np.diff(daily_hours)

print(f"Median: {median_watch_time:.2f}m | Top 10% Threshold (P90): {p90_threshold:.2f}m")
print("Cumulative Hours:", cumulative_watch)
print("Daily Shifts:", np.round(day_to_day_change, 2))
```

```python
# ماتریس امتیازدهی ناقص (حاوی مقادیر خالی)
sparse_ratings = np.array([
    [4.0, np.nan, 3.5],
    [np.nan, 5.0, np.nan],
    [2.0, 3.0, 4.5]
])

# 40. np.isnan: ایجاد ماسک بولی برای شناسایی خانه‌های فاقد امتیاز
nan_mask = np.isnan(sparse_ratings)

# 41. np.nanmean: محاسبه میانگین با نادیده گرفتن مقادیر NaN
safe_user_avg = np.nanmean(sparse_ratings, axis=1)

# 42. np.nan_to_num: پر کردن خودکار خانه‌های خالی با یک مقدار پیش‌فرض (مثلاً 0.0)
imputed_ratings = np.nan_to_num(sparse_ratings, nan=0.0)

# 43. np.copy: ایجاد یک کپی عمیق از آرایه بدون ارجاع اشتراکی در حافظه
isolated_data = np.copy(imputed_ratings)

print("Missing Mask:\n", nan_mask)
print("Safe Mean per User:", safe_user_avg)
print("Imputed Matrix:\n", isolated_data)
```

```python
# 44. np.matmul (یا عملگر @): ضرب ماتریسی پیشرفته برای اعمال فاکتورهای پنهان (Latent Factors)
user_latent = np.array([[1.2, 0.5], [0.3, 2.1]])
item_latent = np.array([[0.9, 1.1], [1.5, 0.2]])
predicted_interactions = np.matmul(user_latent, item_latent.T)

# 45. np.corrcoef: محاسبه ماتریس همبستگی (Pearson Correlation) بین ژانرها
genre_interaction = np.array([[5, 4, 1], [4, 5, 2], [1, 2, 5]])
correlation_matrix = np.corrcoef(genre_interaction)

# 46. np.pad: اضافه کردن حاشیه صفر (Zero Padding) به سیگنال فریم‌ها برای فیلترهای کانوولوشن
signal_slice = np.array([10, 25, 40])
padded_signal = np.pad(signal_slice, pad_width=(1, 1), mode='constant', constant_values=0)

# 47. np.meshgrid: تولید شبکه مختصات دوبعدی برای ارزیابی سطح پاسخ تابع هزینه (Loss Surface)
x_coords = np.linspace(-1, 1, 3)
y_coords = np.linspace(-1, 1, 3)
grid_x, grid_y = np.meshgrid(x_coords, y_coords)

# 48. np.polyfit: برازش رگرسیون خطی روی روند رشد مصرف هفتگی
days = np.array([1, 2, 3, 4, 5])
daily_views = np.array([120, 150, 185, 210, 260])
slope, intercept = np.polyfit(days, daily_views, deg=1)

# 49. np.round / np.floor: گرد کردن خروجی‌ها جهت تبدیل به رتبه‌های عددی گسسته
smooth_predictions = np.round(predicted_interactions, decimals=2)

# 50. np.bincount: شمارش تعداد رخداد هر برچسب دسته‌بندی در بین کاربران
cluster_labels = np.array([0, 1, 1, 2, 0, 1, 1, 2, 2, 2])
cluster_counts = np.bincount(cluster_labels)

print("Correlation Matrix:\n", np.round(correlation_matrix, 2))
print(f"Growth Trend -> Slope: {slope:.2f} views/day, Intercept: {intercept:.2f}")
print("Cluster Frequency Distribution:", cluster_counts)
```
