```python
import numpy as np

# 1. np.array(): تبدیل لیست به آرایه (پایه و اساس کار)
arr = np.array([1, 2, 3, 4, 5, 6])

# 2. np.arange(): ایجاد آرایه با بازه مشخص (شبیه range پایتون)
range_arr = np.arange(0, 10, 2)

# 3. np.linspace(): ایجاد اعداد با فاصله مساوی (بسیار حیاتی برای رسم نمودارها)
linear_space = np.linspace(0, 1, 5)

# 4. np.zeros(): ایجاد آرایه پر از صفر (مناسب برای Initialization وزن‌ها)
zeros = np.zeros((2, 3))

# 5. np.ones(): ایجاد آرایه پر از یک
ones = np.ones((3, 3))

# 6. np.random.rand(): ایجاد اعداد تصادفی در بازه [0, 1)
random_vals = np.random.rand(2, 2)

# 7. np.reshape(): تغییر ابعاد آرایه بدون تغییر داده (مثلاً تبدیل بردار به ماتریس)
reshaped = arr.reshape(2, 3)

# 8. np.flatten(): تبدیل ماتریس چندبعدی به یک بردار تک‌بعدی
flat = reshaped.flatten()

# 9. np.transpose() یا .T: ترانهاده کردن ماتریس (تعویض جای سطر و ستون)
transposed = reshaped.T

# 10. np.concatenate(): چسباندن دو یا چند آرایه به هم
concat = np.concatenate((arr, arr))

# 11. np.split(): تقسیم یک آرایه به چندین زیرآرایه
split_arr = np.split(arr, 2)

# 12. np.argmax(): یافتن شاخص (Index) بزرگترین مقدار (بسیار مهم در خروجی Softmax)
max_idx = np.argmax(arr)

# 13. np.mean(): محاسبه میانگین عناصر
mean_val = np.mean(arr)

# 14. np.std(): محاسبه انحراف معیار (Standard Deviation)
std_dev = np.std(arr)

# 15. np.sum(): مجموع عناصر آرایه
total_sum = np.sum(arr, axis=0)

# 16. np.dot(): ضرب داخلی (Dot Product) یا ضرب ماتریسی
dot_product = np.dot([1, 2], [3, 4])

# 17. np.where(): فیلتر کردن هوشمند (شبیه دستور IF روی آرایه‌ها)
filtered = np.where(arr > 3, 1, 0) # اگر بزرگتر از 3 بود 1 بگذار، وگرنه 0

# 18. np.unique(): یافتن مقادیر منحصر به فرد (مناسب برای دیدن کلاس‌های Labelling)
unique_vals = np.unique(np.array([1, 2, 2, 3, 3, 3]))

# 19. np.expand_dims(): افزودن یک بعد جدید (برای تطبیق ابعاد در مدل‌های Keras)
expanded = np.expand_dims(arr, axis=0)

# 20. np.clip(): محدود کردن مقادیر در یک بازه مشخص (جلوگیری از انفجار گرادیان)
clipped = np.clip(arr, 2, 4)

# نمایش نتایج برای بررسی سریع
print(f"Reshaped:\n{reshaped}\nMax Index: {max_idx}\nFiltered: {filtered}")
```

```python
import numpy as np

# فرض کنیم یک آرایه پایه و دو ماتریس برای عملیات بعدی داریم
arr = np.array([1.5, 2.7, 3.2, 4.9, 5.1])
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

# 21. np.eye(): ایجاد ماتریس همانی (Identity Matrix) - بسیار مهم در جبر خطی
identity = np.eye(3)

# 22. np.diag(): استخراج قطر اصلی یا ساخت ماتریس قطری
diagonal = np.diag([1, 5, 9])

# 23. np.squeeze(): حذف بعدهای تک‌عضوی (مثلاً تبدیل ابعاد (1, 100) به (100,))
squeezed = np.squeeze(np.array([[[1], [2], [3]]]))

# 24. np.vstack(): چسباندن آرایه‌ها به صورت عمودی (Vertical Stack)
v_stacked = np.vstack((matrix_a, matrix_b))

# 25. np.hstack(): چسباندن آرایه‌ها به صورت افقی (Horizontal Stack)
h_stacked = np.hstack((matrix_a, matrix_b))

# 26. np.round_(): گرد کردن عناصر به تعداد ارقام اعشار مشخص
rounded = np.round_(arr, decimals=0)

# 27. np.ceil(): گرد کردن به سمت بالا (سقف)
ceiling = np.ceil(arr)

# 28. np.floor(): گرد کردن به سمت پایین (کف)
floor_vals = np.floor(arr)

# 29. np.abs(): محاسبه قدر مطلق عناصر (بسیار کاربردی در توابع هزینه مثل L1 Loss)
absolute = np.abs(np.array([-1, -2, 3]))

# 30. np.maximum(): مقایسه نظیر به نظیر دو آرایه و انتخاب بیشینه (عملکردی شبیه به ReLU Activation)
relu_like = np.maximum(0, np.array([-2, -1, 3, 5]))

# 31. np.minimum(): مقایسه نظیر به نظیر و انتخاب کمینه
min_compare = np.minimum(matrix_a, matrix_b)

# 32. np.min() / np.max(): یافتن کمترین/بیشترین مقدار در کل آرایه یا در یک راستای خاص
min_val = np.min(matrix_a, axis=1) # کمترین مقدار در هر سطر

# 33. np.percentile(): محاسبه صدک‌ها (مثلاً صدک ۵۰ یا همان میانه برای تشخیص داده‌های پرت Outliers)
p_50 = np.percentile(arr, 50)

# 34. np.var(): محاسبه واریانس داده‌ها (مورد نیاز در فرآیند Standard Scaling)
variance = np.var(arr)

# 35. np.cumsum(): محاسبه مجموع تجمعی عناصر (Cumulative Sum)
cum_sum = np.cumsum(arr)

# 36. np.all(): بررسی اینکه آیا همه عناصر برقرارکننده یک شرط خاص هستند یا خیر (خروجی Boolean)
all_positive = np.all(arr > 0)

# 37. np.any(): بررسی اینکه آیا حداقل یک عنصر برقرارکننده شرط هست یا خیر
any_large = np.any(arr > 4.5)

# 38. np.isnan(): شناسایی داده‌های مفقود شده (NaN) در پیش‌پردازش دیتاست‌ها
has_nan = np.isnan(np.array([1.0, np.nan, 3.0]))

# 39. np.argsort(): بازگرداندن اندیس‌های مرتب‌شده آرایه (پرکاربرد در الگوریتم KNN برای یافتن نزدیک‌ترین همسایه‌ها)
sorted_indices = np.argsort(arr)

# 40. np.tile(): تکرار کردن کل یک آرایه به دفعات مشخص برای ساخت الگوهای تکراری
tiled = np.tile(np.array([1, 2]), (2, 3))

# 41. np.repeat(): تکرار تک‌تک عناصر آرایه به تعداد دفعات مشخص
repeated = np.repeat(np.array([1, 2]), 3)

# 42. np.linalg.inv(): محاسبه معکوس ماتریس (Inverse Matrix) در محاسبات رگرسیون خطی فرمول بسته
matrix_inv = np.linalg.inv(matrix_a)

# 43. np.copy(): ایجاد یک کپی عمیق (Deep Copy) برای جلوگیری از تغییر ناخواسته آرایه اصلی در حافظه
arr_copy = np.copy(arr)

# 44. np.save() / np.load(): ذخیره‌سازی و بازیابی سریع آرایه‌ها با فرمت باینری سبک `.npy`
np.save('[/mnt/data/my_array.npy'](https://storage.gapgpt.app/media/code_interpreter/2f6c19fb-201a-4ebf-b57d-c7917d1762c6/my_array.npy%27), arr)
loaded_arr = np.load('[/mnt/data/my_array.npy'](https://storage.gapgpt.app/media/code_interpreter/2f6c19fb-201a-4ebf-b57d-c7917d1762c6/my_array.npy%27))

# 45. np.random.seed(): تثبیت تولید اعداد تصادفی برای بازتولیدپذیری (Reproducibility) نتایج مدل‌ها
np.random.seed(42)
```
