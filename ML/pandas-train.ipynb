```python
import numpy as np
import pandas as pd

# -------------------------------
# 1) Create a synthetic healthcare dataset (reproducible)
# -------------------------------
rng = np.random.default_rng(42)
n = 500

# Core demographics
age = rng.integers(18, 90, size=n)
sex = rng.choice(["M", "F"], size=n, p=[0.48, 0.52])

# Anthropometrics (with some noise)
height_cm = rng.normal(168, 10, size=n).clip(140, 200)     # realistic bounds
weight_kg = rng.normal(74, 16, size=n).clip(40, 160)

# Vitals / Labs
sbp = rng.normal(125, 18, size=n).clip(85, 220)            # systolic blood pressure
dbp = rng.normal(80, 12, size=n).clip(45, 140)             # diastolic blood pressure
chol_mg_dl = rng.normal(195, 35, size=n).clip(90, 400)     # total cholesterol
glucose_mg_dl = rng.normal(105, 25, size=n).clip(60, 300)  # fasting glucose
smoker = rng.choice([0, 1], size=n, p=[0.75, 0.25])

# A simple synthetic "disease risk" mechanism (probability model)
# This is NOT a medical model; it's for demonstration only.
bmi = weight_kg / (height_cm / 100) * 2
risk_logit = (
    -7.0
    + 0.05  (age - 45)
    + 0.03  (sbp - 120)
    + 0.02  (chol_mg_dl - 180)
    + 0.03  (glucose_mg_dl - 100)
    + 0.8  smoker
    + 0.06 * (bmi - 25)
)

risk_prob = 1 / (1 + np.exp(-risk_logit))
has_cvd = (rng.random(n) < risk_prob).astype(int)  # target label

# Create base DataFrame
df = pd.DataFrame({
    "patient_id": [f"P{100000+i}" for i in range(n)],
    "age": age,
    "sex": sex,
    "height_cm": height_cm.round(1),
    "weight_kg": weight_kg.round(1),
    "sbp": sbp.round(0),
    "dbp": dbp.round(0),
    "cholesterol_mg_dl": chol_mg_dl.round(0),
    "glucose_mg_dl": glucose_mg_dl.round(0),
    "smoker": smoker,
    "has_cvd": has_cvd
})

# -------------------------------
# 2) Inject realistic data issues (missing values, outliers, duplicates)
# -------------------------------
# Missingness
missing_idx = rng.choice(df.index, size=35, replace=False)
df.loc[missing_idx[:10], "cholesterol_mg_dl"] = np.nan
df.loc[missing_idx[10:20], "glucose_mg_dl"] = np.nan
df.loc[missing_idx[20:28], "height_cm"] = np.nan
df.loc[missing_idx[28:], "weight_kg"] = np.nan
```
