```python
import numpy as np

# تنظیم Seed برای بازتولیدپذیری نتایج تصادفی
np.random.seed(42)

# 1. np.array: تعریف ماتریس امتیاز ۴ کاربر به ۳ پلی‌لیست موسیقی
user_ratings = np.array([
    [4.5, 3.0, 5.0],
    [2.0, 1.5, 4.0],
    [5.0, 4.5, 4.5],
    [3.5, 4.0, 2.0]
])

# 2. np.zeros: ساخت بافر خالی برای ثبت لاگ خطاهای سیستم (ابعاد: 4 سرور، 3 وضعیت)
error_buffer = np.zeros((4, 3), dtype=int)

# 3. np.ones: بردار ضرایب اولیه بایاس (Bias weights)
bias_weights = np.ones((3,))

# 4. np.arange: تایم‌لاین شبیه‌سازی تماشای ویدیو (از ثانیه 0 تا 60 با گام 10 ثانیه)
timeline_sec = np.arange(0, 61, 10)

# 5. np.linspace: تولید 5 نقطه زمانی دقیق برای آنالیز فریم‌های کلیدی یک صحنه 2 ثانیه‌ای
frame_timestamps = np.linspace(0.0, 2.0, num=5)

# 6. np.eye: ماتریس همانی برای نگاشت هویت ژانرها به یکدیگر
genre_identity = np.eye(3)

print("Ratings Matrix:\n", user_ratings)
print("Timeline:", timeline_sec)
print("Frames:", frame_timestamps)
```

```python
# 7. np.reshape: تبدیل بردار خطی لاگ استریم به ماتریس (۲ سشن، ۳ کاربر)
raw_stream_logs = np.arange(1, 7)
session_matrix = np.reshape(raw_stream_logs, (2, 3))

# 8. np.transpose (یا .T): ترانهاد ماتریس برای تغییر دیدگاه از (کاربر × پلی‌لیست) به (پلی‌لیست × کاربر)
transposed_ratings = np.transpose(user_ratings)

# 9. np.concatenate: الحاق داده‌های ۲ کاربر جدید به ماتریس اصلی
new_users = np.array([[4.0, 4.0, 4.0], [1.0, 2.0, 3.0]])
all_users_matrix = np.concatenate((user_ratings, new_users), axis=0)

# 10. np.expand_dims: افزودن بُعد Batch Size برای آماده‌سازی ورودی مدل (تبدیل به 3D)
batched_features = np.expand_dims(all_users_matrix, axis=0) # Shape: (1, 6, 3)

# 11. np.squeeze: حذف ابعاد اضافی با طول 1
restored_shape = np.squeeze(batched_features)

print("Concatenated Shape:", all_users_matrix.shape)
print("Batched Shape:", batched_features.shape)
```

```python
# 12. np.mean: میانگین امتیاز داده شده به هر پلی‌لیست (axis=0 ستونی)
avg_playlist_scores = np.mean(user_ratings, axis=0)

# 13. np.std: محاسبه انحراف معیار برای سنجش میزان توافق یا اختلاف نظر کاربران
score_variance = np.std(user_ratings, axis=0)

# 14. np.sum: مجموع زمان صرف‌شده توسط هر کاربر (axis=1 سطری)
total_user_engagement = np.sum(user_ratings, axis=1)

# 15. np.min / np.max: حداقل و حداکثر امتیاز ثبت شده در کل پلتفرم
min_score = np.min(user_ratings)
max_score = np.max(user_ratings)

# 16. np.argmax: شناسایی ایندکس پرطرفدارترین پلی‌لیست بر اساس میانگین
top_playlist_index = np.argmax(avg_playlist_scores)

print(f"Average Scores: {avg_playlist_scores}")
print(f"Top Playlist Index: {top_playlist_index} (Score: {avg_playlist_scores[top_playlist_index]:.2f})")
```

```python
# پروفایل علاقه‌مندی سیستم بر اساس ژانر (وزن‌های تخصیص‌یافته به هر پلی‌لیست)
weights = np.array([0.4, 0.35, 0.25])

# 17. np.dot: ضرب ماتریسی برای محاسبه نمره نهایی شاخص رضایت (User Satisfaction Score)
satisfaction_scores = np.dot(user_ratings, weights)

print(user_ratings)
# 18. np.linalg.norm: (شبیه همون محاسبات فیثاغورثه )محاسبه نرم بردار (طول بردار) برای نرمال‌سازی پروفایل کاربر
user_1_norm = np.linalg.norm(user_ratings[0])
print(user_1_norm)

# 19. np.exp: تابع Softmax ساده‌سازی‌شده برای تبدیل امتیازات به توزیع احتمالاتی ترجیحات
exp_scores = np.exp(user_ratings[0] - np.max(user_ratings[0])) # بهینه از نظر عددی
genre_probabilities = exp_scores / np.sum(exp_scores)

# 20. np.clip: محدود کردن مقادیر خارج از محدوده (مثلاً فیلتر کردن نویزهای فراتر از بازه [1.0, 5.0])
normalized_clamped_ratings = np.clip(user_ratings, a_min=2.0, a_max=4.5)
print("Satisfaction Scores:", satisfaction_scores)
print("Genre Probabilities (User 1):", np.round(genre_probabilities, 3))
```
